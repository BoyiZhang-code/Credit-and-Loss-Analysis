After removing several irrelevant features, a large number of remaining features still contain substantial missing values. Therefore, the next step is to calculate the missing-value ratio for each row.

```python
# Build a dataframe of missing value count and missing value rate

# Define a function to help me check the missing rate
def missing_value_check(data_set):
    # Computing the missing value count for each columns
    missing_count = data_set.isna().sum()

    # computing the missing value rate
    missing_rate = data_set.isna().mean().mul(100)

    # Create a dataframe to contain the columns name, missing value count and missing value rate
    missing_summary = (
        pd.DataFrame({
            'feature_name': data_set.columns,
            'missing_count': missing_count,
            'missing_rate': missing_rate,
        }).sort_values('missing_rate', ascending = False)
    )

    # Filter the missing rate = 0
    missing_summary = missing_summary[missing_summary['missing_count'] > 0]

    # show the table
    return missing_summary
```

```python
# Call the function of missing_value_check to observe the missing rate
missing_value_check(data_clean3)
```

### **05. Observe the distribution of each feature**  
Basic missing-value removal has been completed. The next step is to examine the distribution of each feature and investigate the relationship between candidate features and the target variable.

```python
# Draw the all distribution of each feature
# Draw the histogram for continuous features

# Candidate numerical feature
cols_num = data_clean4.select_dtypes(include=['float64', 'int64']).columns

# Set string-mapped plotting function
plot_map = {
    "hist": sns.histplot,
    "box": sns.boxplot,
    "count": sns.countplot
}

# Define a functions used for drawing different kind of graph with various featuers 
def plot_numeric_dist(df, cols, type_graph, fig_name, n_cols=4):
    """Batch plot univariate distribution graphs for selected features.
    Supports histogram, boxplot and countplot.

    Args:
        df: Input dataset
        cols: List of feature columns to plot (manually separate numeric/categorical columns when calling)
        type_graph: Chart type, accepts "hist", "box", "count"
        fig_name: the chart name in local
        n_cols: Number of subplots per row, default = 4
    """
    
    # Retrieve the corresponding plotting function
    plot_func = plot_map[type_graph]
    
    # Set the layout of distribution graph
    n_rows = (len(cols) + n_cols-1)//n_cols

    # Set the figure size
    plt.figure(figsize=(20, 4 * n_rows))

    # Draw the distribution graph
    for i, col in enumerate(cols):
        plt.subplot(n_rows, n_cols, i+1)
        series = df[col].dropna()

        # IF function is used for chose different type of graphs
        if type_graph == 'hist':
            plot_func(series, kde = True, bins = 30)
        elif type_graph == 'box':
            plot_func(y = series)
        elif type_graph == 'count':
            plot_func(x = series)

        # set the title and lables
        plt.title(f'Distribution of {col}', fontsize=10)
        plt.xlabel(col, fontsize=9)
        plt.ylabel('Count', fontsize=9)

    plt.tight_layout() # Show the graph

    base_path = r"D:\Data_analytics\Creadit and loss analysis\graph\EDA" # set the base path
    save_path = f"{base_path}\\{fig_name}_distribution.png" # set the name of graph
    
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.close()
    # Because of the large of data size, Saving into local can improve the calculation speed
    #plt.show() 
```

```python
# call the function
plot_numeric_dist(data_clean4, cols_num, 'hist', 'numerical')
```

**Categorical vs. Categorical**
In this section, the Cramér’s V matrix is used to examine the strength of association between every pair of categorical features.

```python
# import require library
from scipy.stats import chi2_contingency

# define the function of Cramer's V matrix
def cramer_v(x,y):
    """
    This function calculates Cramer's V coefficient (Yen bias-corrected version).
    Inputs x and y are pandas Series of categorical variables.
    """
    
    contingency_table = pd.crosstab(x,y) # create a table that receive parameters x and y
    chi2 = chi2_contingency(contingency_table)[0] # calculate chi-square
    n = contingency_table.sum().sum() # calculate the sample size
    phi2 = chi2/n # calculate phi2
    r,k = contingency_table.shape # get the number of rows and columns

    # start Yen’s bias correction
    phi2_corr = max(0, phi2 - ((r-1)*(k-1))/(n-1)) # calculate hte phi2_corr
    r_corr = r - ((r-1)**2)/(n-1) # calculate r_corr
    k_corr = k - ((k-1)**2)/(n-1) # calculate k_corr
    
    # identify the sparse situation
    if min((r_corr - 1), (k_corr - 1)) == 0:
        return 0
    return np.sqrt(phi2_corr/min((r_corr - 1), (k_corr - 1))) 
```

The function is defined to compute bias‑corrected Cramer’s V. This coefficient measures the strength of association between two categorical variables (Bergsma, 2013). The Yen correction adjusts for the upward bias of the standard Cramer’s V, especially under small or sparse contingency tables (Bergsma, 2013).

```python
# transform cat_data as dataframe into index
cat_data_index = cat_data.drop(columns = 'sub_grade').columns.tolist()

# build the N×N cramer's V matrix
cramers = pd.DataFrame(index = cat_data_index, columns = cat_data_index, dtype = float)

# fill the value of diagonal
for col in cat_data_index:
    cramers.loc[col, col] = 1.0
    
# Use an upper-triangle loop to construct the association matrix
# which reduces the computation time by half
for i, col1 in enumerate(cat_data_index):
    for col2 in cat_data_index[i + 1:]:
        valid_mask = data_clean6[col1].notna() & data_clean6[col2].notna()
        # calculate the cramer value
        value = cramer_v(data_clean6.loc[valid_mask, col1], 
                         data_clean6.loc[valid_mask, col2])
        # Fill both upper and lower triangle (symmetric matrix)
        cramers.loc[col1, col2] = value
        # copy lower triangle
        cramers.loc[col2, col1] = value
```

```python
# draw the heatmap of relationship between each pair
plt.figure(figsize=(10, 8))
sns.heatmap(cramers, 
            annot=True,          # show the detailed number
            fmt=".2f",           # rounded to 2 decimal places
            cmap='YlGnBu',       # adjust the color
            vmin=0, vmax=1,      # lock the range from 0 - 1 
            linewidths=0.5)

plt.title("Cramer's V Association Matrix (Categorical Features)", fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

Amazing! As the plot shows, most features are independent of one another, with the exception of `loan_status` and `initial_list_status` (the cramer's V is 0.28). From a business perspective, these two variables are indeed related. Therefore, I consider dropping one of them for modelling. This helps avoid multicollinearity and redundant information in our model.

After removing several irrelevant features, a large number of remaining features still contain substantial missing values. Therefore, the next step is to calculate the missing-value ratio for each row.

```python
# Build a dataframe of missing value count and missing value rate

# Define a function to help me check the missing rate
def missing_value_check(data_set):
    # Computing the missing value count for each columns
    missing_count = data_set.isna().sum()

    # computing the missing value rate
    missing_rate = data_set.isna().mean().mul(100)

    # Create a dataframe to contain the columns name, missing value count and missing value rate
    missing_summary = (
        pd.DataFrame({
            'feature_name': data_set.columns,
            'missing_count': missing_count,
            'missing_rate': missing_rate,
        }).sort_values('missing_rate', ascending = False)
    )

    # Filter the missing rate = 0
    missing_summary = missing_summary[missing_summary['missing_count'] > 0]

    # show the table
    return missing_summary
```

# **05. Observe the distribution of each feature**  
Basic missing-value removal has been completed. The next step is to examine the distribution of each feature and investigate the relationship between candidate features and the target variable.

```python
# Draw the all distribution of each feature
# Draw the histogram for continuous features

# Candidate numerical feature
cols_num = data_clean4.select_dtypes(include=['float64', 'int64']).columns

# Set string-mapped plotting function
plot_map = {
    "hist": sns.histplot,
    "box": sns.boxplot,
    "count": sns.countplot
}

# Define a functions used for drawing different kind of graph with various featuers 
def plot_numeric_dist(df, cols, type_graph, fig_name, n_cols=4):
    """Batch plot univariate distribution graphs for selected features.
    Supports histogram, boxplot and countplot.

    Args:
        df: Input dataset
        cols: List of feature columns to plot (manually separate numeric/categorical columns when calling)
        type_graph: Chart type, accepts "hist", "box", "count"
        fig_name: the chart name in local
        n_cols: Number of subplots per row, default = 4
    """
    
    # Retrieve the corresponding plotting function
    plot_func = plot_map[type_graph]
    
    # Set the layout of distribution graph
    n_rows = (len(cols) + n_cols-1)//n_cols

    # Set the figure size
    plt.figure(figsize=(20, 4 * n_rows))

    # Draw the distribution graph
    for i, col in enumerate(cols):
        plt.subplot(n_rows, n_cols, i+1)
        series = df[col].dropna()

        # IF function is used for chose different type of graphs
        if type_graph == 'hist':
            plot_func(series, kde = True, bins = 30)
        elif type_graph == 'box':
            plot_func(y = series)
        elif type_graph == 'count':
            plot_func(x = series)

        # set the title and lables
        plt.title(f'Distribution of {col}', fontsize=10)
        plt.xlabel(col, fontsize=9)
        plt.ylabel('Count', fontsize=9)

    plt.tight_layout() # Show the graph

    base_path = r"D:\Data_analytics\Creadit and loss analysis\graph\EDA" # set the base path
    save_path = f"{base_path}\\{fig_name}_distribution.png" # set the name of graph
    
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.close()
    # Because of the large of data size, Saving into local can improve the calculation speed
    #plt.show() 
```

## **Categorical vs. Categorical**  
In this section, the Cramér’s V matrix is used to examine the strength of association between every pair of categorical features.

```python
# import require library
from scipy.stats import chi2_contingency

# define the function of Cramer's V matrix
def cramer_v(x,y):
    """
    This function calculates Cramer's V coefficient (Yen bias-corrected version).
    Inputs x and y are pandas Series of categorical variables.
    """
    
    contingency_table = pd.crosstab(x,y) # create a table that receive parameters x and y
    chi2 = chi2_contingency(contingency_table)[0] # calculate chi-square
    n = contingency_table.sum().sum() # calculate the sample size
    phi2 = chi2/n # calculate phi2
    r,k = contingency_table.shape # get the number of rows and columns

    # start Yen’s bias correction
    phi2_corr = max(0, phi2 - ((r-1)*(k-1))/(n-1)) # calculate hte phi2_corr
    r_corr = r - ((r-1)**2)/(n-1) # calculate r_corr
    k_corr = k - ((k-1)**2)/(n-1) # calculate k_corr
    
    # identify the sparse situation
    if min((r_corr - 1), (k_corr - 1)) == 0:
        return 0
    return np.sqrt(phi2_corr/min((r_corr - 1), (k_corr - 1))) 
```

The function is defined to compute bias‑corrected Cramer’s V. This coefficient measures the strength of association between two categorical variables (Bergsma, 2013). The Yen correction adjusts for the upward bias of the standard Cramer’s V, especially under small or sparse contingency tables (Bergsma, 2013).

```python
# transform cat_data as dataframe into index
cat_data_index = cat_data.drop(columns = ['sub_grade', 'loan_status']).columns.tolist()

# build the N×N cramer's V matrix
cramers = pd.DataFrame(index = cat_data_index, columns = cat_data_index, dtype = float)

# fill the value of diagonal
for col in cat_data_index:
    cramers.loc[col, col] = 1.0
    
# Use an upper-triangle loop to construct the association matrix
# which reduces the computation time by half
for i, col1 in enumerate(cat_data_index):
    for col2 in cat_data_index[i + 1:]:
        valid_mask = data_clean6[col1].notna() & data_clean6[col2].notna()
        # calculate the cramer value
        value = cramer_v(data_clean6.loc[valid_mask, col1], 
                         data_clean6.loc[valid_mask, col2])
        # Fill both upper and lower triangle (symmetric matrix)
        cramers.loc[col1, col2] = value
        # copy lower triangle
        cramers.loc[col2, col1] = value
```

```python
# draw the heatmap of relationship between each pair
plt.figure(figsize=(10, 8))
sns.heatmap(cramers, 
            annot=True,          # show the detailed number
            fmt=".2f",           # rounded to 2 decimal places
            cmap='YlGnBu',       # adjust the color
            vmin=0, vmax=1,      # lock the range from 0 - 1 
            linewidths=0.5)

plt.title("Cramer's V Association Matrix (Categorical Features)", fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 1. What did I learn about the dataset?

The original Lending Club dataset contains 887,379 observations and 74 features. Through the initial data inspection, I identified that the dataset contains a mixture of borrower information, loan application characteristics, credit bureau records, and repayment-related information.

The dataset is cross-sectional, with one record for each loan rather than repeated monthly repayment snapshots. Therefore, the main objective of this project is to develop an application scorecard (A-card), rather than a behavioural scorecard (B-card) or collection scorecard (C-card).

Several important findings emerged during the EDA:

- **Data availability varies across features and loan vintages.** Some credit bureau variables contain extensive missing values, and their missingness decreases substantially after 2015. - **Numerical features exhibit different distribution patterns.** Several variables are right-skewed, indicating that their distributions should be considered during subsequent feature engineering and scorecard binning.
- **Some numerical features are strongly correlated.** Potentially redundant pairs include `loan_amnt` and `funded_amnt`, `funded_amnt` and `funded_amnt_inv`, `loan_amnt` and `funded_amnt_inv`, as well as `revol_bal` and `total_rev_hi_lim`. - **Categorical features differ in cardinality and data quality.** Some variables contain a manageable number of categories, while others have excessive category counts or inconsistent text values. These characteristics influence the complexity of preprocessing and feature engineering.
- **Numerical distributions vary across certain categorical groups.** The selected boxplots provide preliminary observations of differences in loan amounts, interest rates, income, and DTI across loan status, purpose, and home ownership categories.

Overall, the EDA provides an initial understanding of the dataset's structure, data availability, distribution patterns, and potential relationships between features. These findings will guide subsequent feature selection and modelling decisions.

### Stage 5: Removal of a non-informative numerical feature

After examining the numerical distributions, I checked the values of `policy_code`.

All observations contained the same value, 1.

Since this feature does not distinguish between borrowers, it cannot provide useful discriminatory information in the current dataset.

Therefore, `policy_code` was removed.

The resulting dataset was stored as `data_clean5`.

### Stage 6: Removal of high-cardinality and constant categorical features

After examining the categorical distributions, I identified several features requiring further consideration.

#### `emp_title`

The categorical distribution analysis showed that `emp_title` contains a large number of distinct values and inconsistent text formatting.

The variable also generated most of the font-rendering warnings encountered during categorical plotting.

Given the project's focus on an interpretable credit scorecard and an XGBoost benchmark, I excluded `emp_title` from the initial feature set because of its high cardinality, data quality issues, and preprocessing complexity.

This decision does not establish that employment title information lacks predictive value.

#### `application_type`

After restricting the dataset to individual applications, `application_type` contains only one unique value.

A constant feature cannot distinguish between borrowers and therefore does not provide useful discriminatory information.

Consequently, `application_type` was removed.

The resulting dataset was stored as `data_clean6`.

### Summary of feature-cleaning decisions

The cleaning process was primarily driven by:

1. The intended application-scorecard scope.
2. The availability of features at loan origination.
3. The extremely small proportion of joint applications.
4. Extensive missingness and limited historical data availability.
5. Constant or non-informative features.
6. High-cardinality categorical variables and preprocessing complexity.

The current feature set should be regarded as an initial candidate set rather than a final validated set of predictors. Further feature selection will be conducted during the modelling stage.

## 4. What should be addressed before modelling?

The next stage will focus on defining the modelling objective and preparing a consistent dataset for model development.

### 1. Define the default target

The existing `loan_status` variable contains multiple loan status categories.

Before modelling, I need to define which statuses represent default, which represent non-default, and which observations should be excluded because their repayment outcomes are not sufficiently observed.

The target definition must distinguish completed repayment outcomes from loans that are still ongoing.

### 2. Establish a temporal modelling framework

The dataset contains loans issued across different periods.

Therefore, the modelling process will use a temporal split rather than relying solely on a random train-test split.

This will help evaluate whether the model can generalise to later loan vintages.

### 3. Review missing values and feature availability

The remaining features will be examined to determine:

- Whether missing values represent unavailable information or meaningful business conditions.
- Whether missingness varies across loan vintages.
- Whether the features were available at application origination.
- Whether imputation or missing-value grouping is appropriate.

The modelling process will include:

The scorecard will prioritise interpretability and a clear relationship between borrower characteristics and predicted credit risk.

After removing several irrelevant features, a large number of remaining features still contain substantial missing values. Therefore, the next step is to calculate the missing-value ratio for each row.

```python
# Build a dataframe of missing value count and missing value rate

# Define a function to help me check the missing rate
def missing_value_check(data_set):
    # Computing the missing value count for each columns
    missing_count = data_set.isna().sum()

    # computing the missing value rate
    missing_rate = data_set.isna().mean().mul(100)

    # Create a dataframe to contain the columns name, missing value count and missing value rate
    missing_summary = (
        pd.DataFrame({
            'feature_name': data_set.columns,
            'missing_count': missing_count,
            'missing_rate': missing_rate,
        }).sort_values('missing_rate', ascending = False)
    )

    # Filter the missing rate = 0
    missing_summary = missing_summary[missing_summary['missing_count'] > 0]

    # show the table
    return missing_summary
```

```python
# Draw the all distribution of each feature
# Draw the histogram for continuous features

# Candidate numerical feature
cols_num = data_clean4.select_dtypes(include=['float64', 'int64']).columns

# Set string-mapped plotting function
plot_map = {
    "hist": sns.histplot,
    "box": sns.boxplot,
    "count": sns.countplot
}

# Define a functions used for drawing different kind of graph with various featuers 
def plot_numeric_dist(df, cols, type_graph, fig_name, n_cols=4):
    """Batch plot univariate distribution graphs for selected features.
    Supports histogram, boxplot and countplot.

    Args:
        df: Input dataset
        cols: List of feature columns to plot (manually separate numeric/categorical columns when calling)
        type_graph: Chart type, accepts "hist", "box", "count"
        fig_name: the chart name in local
        n_cols: Number of subplots per row, default = 4
    """
    
    # Retrieve the corresponding plotting function
    plot_func = plot_map[type_graph]
    
    # Set the layout of distribution graph
    n_rows = (len(cols) + n_cols-1)//n_cols

    # Set the figure size
    plt.figure(figsize=(20, 4 * n_rows))

    # Draw the distribution graph
    for i, col in enumerate(cols):
        plt.subplot(n_rows, n_cols, i+1)
        series = df[col].dropna()

        # IF function is used for chose different type of graphs
        if type_graph == 'hist':
            plot_func(series, kde = True, bins = 30)
        elif type_graph == 'box':
            plot_func(y = series)
        elif type_graph == 'count':
            plot_func(x = series)

        # set the title and lables
        plt.title(f'Distribution of {col}', fontsize=10)
        plt.xlabel(col, fontsize=9)
        plt.ylabel('Count', fontsize=9)

    plt.tight_layout() # Show the graph

    base_path = r"D:\Data_analytics\Creadit and loss analysis\graph\EDA" # set the base path
    save_path = f"{base_path}\\{fig_name}_distribution.png" # set the name of graph
    
    plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.close()
    # Because of the large of data size, Saving into local can improve the calculation speed
    #plt.show() 
```

## **Categorical vs. Categorical**  
In this section, the Cramér’s V matrix is used to examine the strength of association between every pair of categorical features.

```python
# import require library
from scipy.stats import chi2_contingency

# define the function of Cramer's V matrix
def cramer_v(x,y):
    """
    This function calculates Cramer's V coefficient (Yen bias-corrected version).
    Inputs x and y are pandas Series of categorical variables.
    """
    
    contingency_table = pd.crosstab(x,y) # create a table that receive parameters x and y
    chi2 = chi2_contingency(contingency_table)[0] # calculate chi-square
    n = contingency_table.sum().sum() # calculate the sample size
    phi2 = chi2/n # calculate phi2
    r,k = contingency_table.shape # get the number of rows and columns

    # start Yen’s bias correction
    phi2_corr = max(0, phi2 - ((r-1)*(k-1))/(n-1)) # calculate hte phi2_corr
    r_corr = r - ((r-1)**2)/(n-1) # calculate r_corr
    k_corr = k - ((k-1)**2)/(n-1) # calculate k_corr
    
    # identify the sparse situation
    if min((r_corr - 1), (k_corr - 1)) == 0:
        return 0
    return np.sqrt(phi2_corr/min((r_corr - 1), (k_corr - 1))) 
```

The function is defined to compute bias‑corrected Cramer’s V. This coefficient measures the strength of association between two categorical variables (Bergsma, 2013). The Yen correction adjusts for the upward bias of the standard Cramer’s V, especially under small or sparse contingency tables (Bergsma, 2013).

```python
# transform cat_data as dataframe into index
cat_data_index = cat_data.drop(columns = ['sub_grade', 'loan_status']).columns.tolist()

# build the N×N cramer's V matrix
cramers = pd.DataFrame(index = cat_data_index, columns = cat_data_index, dtype = float)

# fill the value of diagonal
for col in cat_data_index:
    cramers.loc[col, col] = 1.0
    
# Use an upper-triangle loop to construct the association matrix
# which reduces the computation time by half
for i, col1 in enumerate(cat_data_index):
    for col2 in cat_data_index[i + 1:]:
        valid_mask = data_clean6[col1].notna() & data_clean6[col2].notna()
        # calculate the cramer value
        value = cramer_v(data_clean6.loc[valid_mask, col1], 
                         data_clean6.loc[valid_mask, col2])
        # Fill both upper and lower triangle (symmetric matrix)
        cramers.loc[col1, col2] = value
        # copy lower triangle
        cramers.loc[col2, col1] = value
```

```python
# draw the heatmap of relationship between each pair
plt.figure(figsize=(10, 8))
sns.heatmap(cramers, 
            annot=True,          # show the detailed number
            fmt=".2f",           # rounded to 2 decimal places
            cmap='YlGnBu',       # adjust the color
            vmin=0, vmax=1,      # lock the range from 0 - 1 
            linewidths=0.5)

plt.title("Cramer's V Association Matrix (Categorical Features)", fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

- `zip_code`: Granular geographic information with high-cardinality complexity.

**Stage 2 – Mid-loan and post-loan feature removal (`data_clean2`)**

- `collection_recovery_fee`, `last_credit_pull_d`, `last_pymnt_amnt`, `last_pymnt_d`, `next_pymnt_d`, `out_prncp`, `out_prncp_inv`, `pymnt_plan`, `recoveries`, `total_pymnt`, `total_pymnt_inv`, `total_rec_int`, `total_rec_late_fee`, `total_rec_prncp`: Removed because they relate to loan monitoring, repayment, outstanding balances, or collection outcomes and are not appropriate as application-time predictors.
- `term`: Also removed during this stage; its exclusion rationale requires further review because it is an application-level feature.

**Stage 3 – Joint application removal (`data_clean3`)**

- `annual_inc_joint`, `dti_joint`, `verification_status_joint`: Joint-borrower-specific features.
- Joint applications were excluded to focus on individual borrowers and maintain a consistent modelling population. They account for approximately 0.06% of observations.

**Stage 4 – High-missingness feature removal (`data_clean4`)**

- `il_util`, `mths_since_rcnt_il`, `open_rv_12m`, `open_acc_6m`, `open_il_6m`, `open_il_12m`, `open_il_24m`, `total_bal_il`, `inq_last_12m`, `total_cu_tl`, `open_rv_24m`, `max_bal_bc`, `all_util`, `inq_fi`.
- Removed due to extensive missingness (over 90%) and limited historical availability, which would complicate the development of a consistent model across loan vintages.

**Stage 5 – Non-informative feature removal (`data_clean5`)**

- `policy_code`: Removed because all observations contain the same value (1), providing no discriminatory information.

**Stage 6 – High-cardinality and constant feature removal (`data_clean6`)**

- `emp_title`: Excluded due to excessive unique categories, inconsistent text formatting, and preprocessing complexity.
- `application_type`: Removed because it becomes constant after restricting the dataset to individual applications.

The resulting `data_clean6` represents the initial cleaned feature set. These exclusions reflect project scope and preprocessing considerations rather than confirmed lack of predictive value, and the remaining features will be further assessed during modelling.

## 3. Which features remain as candidate predictors?

The remaining features cover several major categories:

- **Loan characteristics:** `loan_amnt`, `funded_amnt`, `funded_amnt_inv`, `int_rate`, `installment`.
- **Borrower characteristics:** `emp_length`, `home_ownership`, `annual_inc`.
- **Application information:** `verification_status`, `purpose`, `addr_state`, `issue_d`.
- **Credit history:** `delinq_2yrs`, `inq_last_6mths`, `mths_since_last_delinq`, `mths_since_last_record`, `open_acc`, `pub_rec`, `revol_bal`, `revol_util`, `total_acc`.
- **Other structured features:** `grade`, `sub_grade`, `initial_list_status`.

These features form the initial candidate pool rather than a final predictor set. Correlation, missingness, data availability at origination, encoding requirements, and predictive contribution will be further assessed during modelling.

## 4. What should be addressed before modelling?

The next stage will focus on:

1. **Target definition:** Define default and non-default loan statuses, excluding loans with insufficiently observed outcomes.
2. **Temporal split:** Establish training, validation, and out-of-time test periods to evaluate performance on later loan vintages.
3. **Feature preparation:** Review missingness, origination-time availability, binning, encoding, and potential redundancy.
4. **Scorecard development:** Apply binning, WoE, IV, and Logistic Regression to develop an interpretable credit scorecard.
5. **XGBoost benchmark:** Develop a challenger model using the same target and temporal evaluation framework.
6. **Model evaluation:** Compare discrimination, calibration, and stability using ROC-AUC, PR-AUC, KS, Brier Score, Log Loss, calibration analysis, and temporal/segment stability.

## Overall conclusion

The EDA established an initial cleaning framework and identified key data characteristics, including historical feature availability, distribution patterns, potential redundancy, and categorical complexity. The cleaned dataset is ready for target definition and modelling preparation, while the final feature set will be determined through further evaluation.

The next step is to define the default target and temporal split, followed by development of the interpretable credit scorecard and its XGBoost benchmark.
